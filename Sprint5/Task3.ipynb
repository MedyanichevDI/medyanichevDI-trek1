```
using System;
using System.Collections.Generic;
using System.Linq;


// ИНТЕРФЕЙСЫ


public interface IPayable
{
    double CalculatePayment(double amount);
    string GetSummary();
}

public interface IShippable
{
    double CalculateShippingCost(double weightKg);
    string GetSummary();
}

public interface ICreditable
{
    double CreditLimit { get; set; }
    bool IsCreditAvailable(double amount);
}

 
// ДЕЛЕГАТЫ
 

public delegate bool SupplierFilter(Supplier supplier);
public delegate TResult SupplierProjector<in T, out TResult>(T supplier);
public delegate void RatingChangedHandler(Supplier supplier, double oldRating, double newRating);
public delegate void OrderPlacedHandler(Supplier supplier, string orderDetails, double total);
public delegate double DiscountCalculator(double basePrice, Supplier supplier);

 
// БАЗОВЫЙ КЛАСС
 

public class Supplier : IPayable, IShippable, ICreditable
{
    private int _supplierId;
    private string _companyName;
    private string _productType;
    private double _rating;
    private bool _isActive;
    private string _contactEmail;
    private int _totalOrders;

    //   4 новых атрибута  
    private int _paymentTermsDays;
    private double _creditLimit;
    private DateTime _registrationDate;
    private string _bankAccount;

    //   НОВЫЕ АТРИБУТЫ (коллекции)  
    private readonly List<string> _tags = new List<string>();
    private readonly Dictionary<string, string> _contacts = new Dictionary<string, string>();
    private readonly Queue<string> _orderHistory = new Queue<string>();
    private readonly HashSet<string> _certifications = new HashSet<string>();

    public int SupplierId { get => _supplierId; set => _supplierId = value; }

    public string CompanyName
    {
        get => _companyName;
        set
        {
            if (string.IsNullOrEmpty(value))
                throw new ArgumentException("Название компании не может быть пустым!");
            _companyName = value;
        }
    }

    public string ProductType { get => _productType; protected set => _productType = value; }

    public double Rating
    {
        get => _rating;
        protected set
        {
            if (value < 0 || value > 5)
                throw new ArgumentException("Рейтинг должен быть в диапазоне от 0 до 5!");
            _rating = value;
        }
    }

    public bool IsActive { get => _isActive; set => _isActive = value; }

    public string ContactEmail
    {
        get => _contactEmail;
        set
        {
            if (value == null || !value.Contains("@"))
                throw new ArgumentException("Некорректный email адрес!");
            _contactEmail = value;
        }
    }

    public int TotalOrders { get => _totalOrders; private set => _totalOrders = value; }

    //   4 новых свойства  
    public int PaymentTermsDays
    {
        get => _paymentTermsDays;
        set
        {
            if (value < 0)
                throw new ArgumentException("Срок оплаты не может быть отрицательным!");
            _paymentTermsDays = value;
        }
    }

    public double CreditLimit
    {
        get => _creditLimit;
        set
        {
            if (value < 0)
                throw new ArgumentException("Кредитный лимит не может быть отрицательным!");
            _creditLimit = value;
        }
    }

    public DateTime RegistrationDate { get => _registrationDate; protected set => _registrationDate = value; }

    public string BankAccount
    {
        get => _bankAccount;
        set
        {
            if (value == null || value.Length < 5)
                throw new ArgumentException("Некорректный номер банковского счёта!");
            _bankAccount = value;
        }
    }

    //   НОВЫЕ СВОЙСТВА (коллекции)  
    public IReadOnlyList<string> Tags => _tags;
    public IReadOnlyDictionary<string, string> Contacts => _contacts;
    public int OrderHistoryCount => _orderHistory.Count;
    public IReadOnlyCollection<string> Certifications => _certifications;

    //   НОВЫЕ СОБЫТИЯ  
    public event RatingChangedHandler RatingChanged;
    public event OrderPlacedHandler OrderPlaced;
    public event Action<Supplier, string> SupplierDeactivated;

    //   PROTECTED-ОБЁРТКИ ДЛЯ ВЫЗОВА СОБЫТИЙ  
    protected void RaiseRatingChanged(double oldRating, double newRating)
        => RatingChanged?.Invoke(this, oldRating, newRating);

    protected void RaiseOrderPlaced(string orderDetails, double total)
        => OrderPlaced?.Invoke(this, orderDetails, total);

    protected void RaiseSupplierDeactivated(string reason)
        => SupplierDeactivated?.Invoke(this, reason);

    public Supplier(int supplierId, string companyName, string productType, string contactEmail)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        ContactEmail = contactEmail;
        Rating = 0;
        IsActive = true;
        TotalOrders = 0;
        PaymentTermsDays = 30;
        CreditLimit = 0;
        RegistrationDate = DateTime.Now;
        BankAccount = "0000000000";
    }

    // Перегрузка: два метода ProvideQuote
    public virtual string ProvideQuote()
        => $"Котировка от {CompanyName}: стандартные условия поставки {ProductType}.";

    public virtual string ProvideQuote(int quantity)
        => $"Котировка от {CompanyName} на партию {quantity} ед. {ProductType}.";

    public virtual void SendQuote() => Console.WriteLine(ProvideQuote());

    // Перегрузка: два метода SubmitOrder
    public virtual string SubmitOrder(string orderDetails)
    {
        RegisterOrder();
        PushOrderToHistory(orderDetails);
        RaiseOrderPlaced(orderDetails, 0);
        return $"Заказ '{orderDetails}' отправлен поставщику {CompanyName} (ID: {SupplierId}).";
    }

    public virtual string SubmitOrder(string orderDetails, DateTime deliveryDate)
    {
        RegisterOrder();
        PushOrderToHistory(orderDetails);
        RaiseOrderPlaced(orderDetails, 0);
        return $"Заказ '{orderDetails}' отправлен поставщику {CompanyName}, доставка: {deliveryDate:dd.MM.yyyy}.";
    }

    public virtual string GetCompanyInfo()
        => $"ID: {SupplierId}, Компания: {CompanyName}, Тип продукции: {ProductType}, Рейтинг: {Rating}";

    public void UpdateRating(double newRating)
    {
        double old = Rating;
        Rating = newRating;
        RaiseRatingChanged(old, newRating);
        Console.WriteLine($"Рейтинг {CompanyName} обновлён: {Rating}");
    }

    public void Deactivate()
    {
        IsActive = false;
        RaiseSupplierDeactivated("Деактивирован вручную");
        Console.WriteLine($"{CompanyName} деактивирован.");
    }

    public void LinkWithSupplier(Supplier other)
        => Console.WriteLine($"Поставщик {CompanyName} (ID: {SupplierId}) связан с {other.CompanyName} (ID: {other.SupplierId})");

    protected void RegisterOrder() => TotalOrders++;

    //   4 новых метода  
    public void ExtendPaymentTerms(int extraDays)
    {
        PaymentTermsDays += extraDays;
        Console.WriteLine($"Срок оплаты {CompanyName} продлён до {PaymentTermsDays} дн.");
    }

    public int GetYearsOnMarket() => DateTime.Now.Year - RegistrationDate.Year;

    public bool IsCreditAvailable(double amount) => amount <= CreditLimit;

    public void Audit(string inspectorName)
        => Console.WriteLine($"[Аудит] {inspectorName} проверил {CompanyName}: активен={IsActive}, заказов={TotalOrders}");

    //   НОВЫЕ МЕТОДЫ (коллекции, делегаты)  
    public void AddTag(string tag)
    {
        if (!string.IsNullOrWhiteSpace(tag) && !_tags.Contains(tag))
        {
            _tags.Add(tag);
            Console.WriteLine($"Тег '{tag}' добавлен поставщику {CompanyName}.");
        }
    }

    public bool HasTag(string tag) => _tags.Contains(tag);

    public void AddContact(string role, string value)
    {
        if (string.IsNullOrWhiteSpace(role) || string.IsNullOrWhiteSpace(value))
            throw new ArgumentException("Роль и значение контакта не могут быть пустыми!");
        _contacts[role] = value;
        Console.WriteLine($"Контакт '{role}' = '{value}' добавлен для {CompanyName}.");
    }

    public string GetContact(string role) => _contacts.TryGetValue(role, out var v) ? v : null;

    public void PushOrderToHistory(string orderDetails)
    {
        _orderHistory.Enqueue(orderDetails);
        while (_orderHistory.Count > 20) _orderHistory.Dequeue();
    }

    public string PeekLastOrder() => _orderHistory.Count > 0 ? _orderHistory.Peek() : null;

    public List<string> DrainOrderHistory()
    {
        var result = new List<string>();
        while (_orderHistory.Count > 0) result.Add(_orderHistory.Dequeue());
        return result;
    }

    public bool AddCertification(string cert)
    {
        bool added = _certifications.Add(cert);
        Console.WriteLine(added
            ? $"Сертификат '{cert}' добавлен {CompanyName}."
            : $"Сертификат '{cert}' уже есть у {CompanyName}.");
        return added;
    }

    public static List<Supplier> FilterSuppliers(IEnumerable<Supplier> source, SupplierFilter filter)
    {
        var result = new List<Supplier>();
        foreach (var s in source)
            if (filter(s)) result.Add(s);
        return result;
    }

    public static List<TResult> ProjectSuppliers<TResult>(
        IEnumerable<Supplier> source, SupplierProjector<Supplier, TResult> projector)
    {
        var result = new List<TResult>();
        foreach (var s in source) result.Add(projector(s));
        return result;
    }

    public double CalculateDiscountedPrice(double basePrice, DiscountCalculator calculator)
        => calculator(basePrice, this);

    // ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСОВ
    double IPayable.CalculatePayment(double amount) => CalculatePaymentCore(amount);
    string IPayable.GetSummary() => $"[Оплата] {CompanyName}: отсрочка {PaymentTermsDays} дн., лимит {CreditLimit:F0}";
    double IShippable.CalculateShippingCost(double weightKg) => CalculateShippingCore(weightKg);
    string IShippable.GetSummary() => $"[Доставка] {CompanyName}: тариф рассчитывается по весу груза";

    protected virtual double CalculatePaymentCore(double amount) => amount;
    protected virtual double CalculateShippingCore(double weightKg) => weightKg * 10;
}

 
// MANUFACTURER
 

public class Manufacturer : Supplier
{
    private int _foundedYear;
    private int _leadTimeDays;
    private int _minOrderQuantity;
    private int _warrantyMonths;
    private string _certificationBody;

    //   НОВЫЕ АТРИБУТЫ  
    private readonly Dictionary<string, double> _componentCosts = new Dictionary<string, double>();
    private readonly List<string> _productionLines = new List<string>();
    private readonly HashSet<string> _qualityChecks = new HashSet<string>();
    private int _maxLineLoad = 100;

    public int FoundedYear
    {
        get => _foundedYear;
        set
        {
            if (value <= 1800 || value > DateTime.Now.Year)
                throw new ArgumentException("Некорректный год основания!");
            _foundedYear = value;
        }
    }

    public int ProductionCapacity { get; set; }
    public string QualityCertificate { get; set; }
    public double DefectRate { get; set; }

    public int LeadTimeDays
    {
        get => _leadTimeDays;
        set
        {
            if (value < 0) throw new ArgumentException("Срок производства не может быть отрицательным!");
            _leadTimeDays = value;
        }
    }

    public int MinOrderQuantity
    {
        get => _minOrderQuantity;
        set
        {
            if (value < 0) throw new ArgumentException("Мин. партия не может быть отрицательной!");
            _minOrderQuantity = value;
        }
    }

    public int WarrantyMonths
    {
        get => _warrantyMonths;
        set
        {
            if (value < 0) throw new ArgumentException("Гарантия не может быть отрицательной!");
            _warrantyMonths = value;
        }
    }

    public string CertificationBody { get => _certificationBody; set => _certificationBody = value; }

    //   НОВЫЕ СВОЙСТВА  
    public IReadOnlyDictionary<string, double> ComponentCosts => _componentCosts;
    public IReadOnlyList<string> ProductionLines => _productionLines;
    public int MaxLineLoad { get => _maxLineLoad; set => _maxLineLoad = value; }

    public Manufacturer(int supplierId, string companyName, string productType, string contactEmail, int foundedYear)
        : base(supplierId, companyName, productType, contactEmail)
    {
        FoundedYear = foundedYear;
    }

    public override string ProvideQuote()
        => $"Котировка от производителя {CompanyName} (основан в {FoundedYear} г.): специальные условия для {ProductType}.";

    public override string ProvideQuote(int quantity)
    {
        string note = quantity > ProductionCapacity ? " (требуется доп. время на изготовление)" : "";
        return $"Котировка от производителя {CompanyName} на {quantity} ед. {ProductType}{note}.";
    }

    public void SupplyToRetailer(Retailer other, string productDetails)
        => Console.WriteLine($"Производитель {CompanyName} поставляет '{productDetails}' ритейлеру {other.CompanyName} ({other.StoreLocation})");

    public bool CheckCapacity(int requestedQuantity) => requestedQuantity <= ProductionCapacity;

    //   4 новых метода  
    public bool IsOrderValid(int quantity) => quantity >= MinOrderQuantity;

    public int EstimateProductionTime(int quantity)
    {
        if (ProductionCapacity <= 0) return LeadTimeDays;
        int extraMonths = (int)Math.Ceiling((double)quantity / ProductionCapacity) - 1;
        return LeadTimeDays + extraMonths * 30;
    }

    public double CalculateWarrantyCost(double basePrice) => basePrice * (WarrantyMonths / 12.0) * 0.02;

    public void PassCertification(string newCertificate, string body)
    {
        QualityCertificate = newCertificate;
        CertificationBody = body;
        Console.WriteLine($"{CompanyName} прошёл сертификацию: {newCertificate} ({body})");
    }

    //   НОВЫЕ МЕТОДЫ  
    public void AddProductionLine(string lineName)
    {
        if (!string.IsNullOrWhiteSpace(lineName) && !_productionLines.Contains(lineName))
        {
            _productionLines.Add(lineName);
            Console.WriteLine($"Линия '{lineName}' добавлена производителю {CompanyName}.");
        }
    }

    public void SetComponentCost(string component, double cost)
    {
        if (cost < 0) throw new ArgumentException("Стоимость не может быть отрицательной!");
        _componentCosts[component] = cost;
    }

    public double GetTotalComponentCost() => _componentCosts.Values.Sum();

    public bool PassQualityCheck(string checkName)
    {
        bool added = _qualityChecks.Add(checkName);
        Console.WriteLine(added
            ? $"Проверка '{checkName}' пройдена {CompanyName}."
            : $"Проверка '{checkName}' уже пройдена.");
        return added;
    }

    public bool IsQualityCheckPassed(string checkName) => _qualityChecks.Contains(checkName);

    public double EstimateLineLoad(int quantity)
    {
        if (ProductionCapacity <= 0) return 100;
        return Math.Min(_maxLineLoad, quantity * 100.0 / ProductionCapacity);
    }

    // Скидка 3% при короткой отсрочке платежа
    protected override double CalculatePaymentCore(double amount) => PaymentTermsDays <= 14 ? amount * 0.97 : amount;
}

 
// RETAILER
 

public class Retailer : Supplier
{
    private string _storeLocation;
    private int _returnPolicyDays;
    private bool _hasOnlineStore;
    private double _averageCheck;
    private bool _loyaltyProgram;

    //   НОВЫЕ АТРИБУТЫ  
    private readonly Dictionary<string, int> _stock = new Dictionary<string, int>();
    private readonly List<string> _promotions = new List<string>();
    private readonly HashSet<string> _loyaltyMembers = new HashSet<string>();
    private double _cashbackRate = 0.03;

    public string StoreLocation
    {
        get => _storeLocation;
        set
        {
            if (string.IsNullOrEmpty(value))
                throw new ArgumentException("Местоположение магазина не может быть пустым!");
            _storeLocation = value;
        }
    }

    public int EmployeeCount { get; set; }
    public string OpeningHours { get; set; }
    public double DiscountPercentage { get; set; }

    public int ReturnPolicyDays
    {
        get => _returnPolicyDays;
        set
        {
            if (value < 0) throw new ArgumentException("Срок возврата не может быть отрицательным!");
            _returnPolicyDays = value;
        }
    }

    public bool HasOnlineStore { get => _hasOnlineStore; set => _hasOnlineStore = value; }

    public double AverageCheck
    {
        get => _averageCheck;
        set
        {
            if (value < 0) throw new ArgumentException("Средний чек не может быть отрицательным!");
            _averageCheck = value;
        }
    }

    public bool LoyaltyProgram { get => _loyaltyProgram; set => _loyaltyProgram = value; }

    //   НОВЫЕ СВОЙСТВА  
    public IReadOnlyDictionary<string, int> Stock => _stock;
    public IReadOnlyList<string> Promotions => _promotions;
    public int LoyaltyMembersCount => _loyaltyMembers.Count;
    public double CashbackRate { get => _cashbackRate; set => _cashbackRate = value; }

    public Retailer(int supplierId, string companyName, string productType, string contactEmail, string storeLocation)
        : base(supplierId, companyName, productType, contactEmail)
    {
        StoreLocation = storeLocation;
    }

    public override string SubmitOrder(string orderDetails)
    {
        RegisterOrder();
        PushOrderToHistory(orderDetails);
        RaiseOrderPlaced(orderDetails, 0);
        return $"Заказ '{orderDetails}' отправлен ритейлеру {CompanyName} (магазин: {StoreLocation}).";
    }

    public override string SubmitOrder(string orderDetails, DateTime deliveryDate)
    {
        RegisterOrder();
        PushOrderToHistory(orderDetails);
        RaiseOrderPlaced(orderDetails, 0);
        return $"Заказ '{orderDetails}' отправлен ритейлеру {CompanyName} (магазин: {StoreLocation}), к {deliveryDate:dd.MM.yyyy}.";
    }

    public void OrderFromManufacturer(Manufacturer other, string orderDetails)
    {
        Console.WriteLine($"Ритейлер {CompanyName} заказывает '{orderDetails}' у производителя {other.CompanyName}");
        Console.WriteLine(other.SubmitOrder(orderDetails));
    }

    public void OrderFromImporter(Importer other, string orderDetails)
    {
        Console.WriteLine($"Ритейлер {CompanyName} заказывает '{orderDetails}' у импортёра {other.CompanyName}");
        Console.WriteLine(other.SubmitOrder(orderDetails));
    }

    public double ApplyDiscount(double price) => price - price * DiscountPercentage / 100;

    //   4 новых метода  
    public bool CanReturn(int daysSincePurchase) => daysSincePurchase >= 0 && daysSincePurchase <= ReturnPolicyDays;

    public double EstimateCheckTotal(int itemsCount) => ApplyDiscount(AverageCheck * itemsCount);

    public double CalculateLoyaltyBonus(double purchaseTotal) => LoyaltyProgram ? purchaseTotal * 0.05 : 0;

    public void ToggleOnlineStore()
    {
        HasOnlineStore = !HasOnlineStore;
        Console.WriteLine($"Онлайн-магазин {CompanyName}: {(HasOnlineStore ? "включён" : "выключен")}");
    }

    //   НОВЫЕ МЕТОДЫ  
    public void AddStock(string product, int quantity)
    {
        if (quantity < 0) throw new ArgumentException("Количество не может быть отрицательным!");
        _stock[product] = _stock.TryGetValue(product, out var cur) ? cur + quantity : quantity;
        Console.WriteLine($"Склад {CompanyName}: '{product}' = {_stock[product]} шт.");
    }

    public bool SellProduct(string product, int quantity)
    {
        if (!_stock.TryGetValue(product, out var cur) || cur < quantity)
        {
            Console.WriteLine($"Недостаточно '{product}' на складе {CompanyName}.");
            return false;
        }
        _stock[product] = cur - quantity;
        Console.WriteLine($"Продано {quantity} шт. '{product}' из {CompanyName}. Остаток: {_stock[product]}.");
        return true;
    }

    public void AddPromotion(string promo)
    {
        if (!string.IsNullOrWhiteSpace(promo) && !_promotions.Contains(promo))
        {
            _promotions.Add(promo);
            Console.WriteLine($"Акция '{promo}' запущена в {CompanyName}.");
        }
    }

    public bool RegisterLoyaltyMember(string email)
    {
        bool added = _loyaltyMembers.Add(email);
        Console.WriteLine(added
            ? $"Клиент {email} зарегистрирован в программе лояльности {CompanyName}."
            : $"Клиент {email} уже участвует в программе.");
        return added;
    }

    public double CalculateCashback(double purchaseTotal, DiscountCalculator custom = null)
    {
        if (custom != null) return custom(purchaseTotal, this);
        return purchaseTotal * _cashbackRate;
    }

    // Доставка по тарифу 8 за кг, но не менее 100
    protected override double CalculateShippingCore(double weightKg) => Math.Max(100, weightKg * 8);
}

 
// IMPORTER
 

public class Importer : Supplier
{
    private string _importLicenseNumber;
    private int _transitDays;
    private double _insuranceRate;
    private string _portOfEntry;

    //   НОВЫЕ АТРИБУТЫ  
    private readonly Dictionary<string, double> _customsDuties = new Dictionary<string, double>();
    private readonly Queue<string> _shipmentQueue = new Queue<string>();
    private readonly HashSet<string> _visitedPorts = new HashSet<string>();
    private string _preferredCarrier = "Maersk";

    public string OriginCountry { get; set; }
    public string CustomsCode { get; set; }
    public string ShippingMethod { get; set; }
    public double ImportTax { get; set; }

    public string ImportLicenseNumber { get => _importLicenseNumber; set => _importLicenseNumber = value; }

    public int TransitDays
    {
        get => _transitDays;
        set
        {
            if (value < 0) throw new ArgumentException("Срок транзита не может быть отрицательным!");
            _transitDays = value;
        }
    }

    public double InsuranceRate
    {
        get => _insuranceRate;
        set
        {
            if (value < 0 || value > 100) throw new ArgumentException("Страховка должна быть от 0 до 100%!");
            _insuranceRate = value;
        }
    }

    public string PortOfEntry { get => _portOfEntry; set => _portOfEntry = value; }

    //   НОВЫЕ СВОЙСТВА  
    public IReadOnlyDictionary<string, double> CustomsDuties => _customsDuties;
    public int ShipmentQueueCount => _shipmentQueue.Count;
    public IReadOnlyCollection<string> VisitedPorts => _visitedPorts;
    public string PreferredCarrier { get => _preferredCarrier; set => _preferredCarrier = value; }

    public Importer(int supplierId, string companyName, string productType, string contactEmail, string originCountry)
        : base(supplierId, companyName, productType, contactEmail)
    {
        OriginCountry = originCountry;
    }

    public override string GetCompanyInfo()
        => base.GetCompanyInfo() + $", Страна: {OriginCountry}, пошлина: {ImportTax}%";

    public void SupplyToManufacturer(Manufacturer other, string componentDetails)
        => Console.WriteLine($"Импортёр {CompanyName} поставляет '{componentDetails}' производителю {other.CompanyName} (из: {OriginCountry})");

    public double CalculateTotalCost(double baseCost) => baseCost + baseCost * ImportTax / 100;

    //   4 новых метода  
    public double CalculateInsurance(double cargoValue) => cargoValue * InsuranceRate / 100;

    public DateTime EstimateDelivery(DateTime orderDate) => orderDate.AddDays(TransitDays);

    public bool IsLicenseValid() => !string.IsNullOrEmpty(ImportLicenseNumber);

    public void UpdatePortOfEntry(string newPort)
    {
        PortOfEntry = newPort;
        Console.WriteLine($"Порт назначения {CompanyName} изменён на {newPort}");
    }

    //   НОВЫЕ МЕТОДЫ  
    public void SetCustomsDuty(string category, double percent)
    {
        if (percent < 0 || percent > 100) throw new ArgumentException("Пошлина 0..100%!");
        _customsDuties[category] = percent;
        Console.WriteLine($"Пошлина '{category}' для {CompanyName} = {percent}%.");
    }

    public double GetCustomsDuty(string category)
        => _customsDuties.TryGetValue(category, out var p) ? p : ImportTax;

    public void EnqueueShipment(string shipment)
    {
        _shipmentQueue.Enqueue(shipment);
        Console.WriteLine($"Поставка '{shipment}' добавлена в очередь {CompanyName} (всего: {_shipmentQueue.Count}).");
    }

    public string ProcessNextShipment()
    {
        if (_shipmentQueue.Count == 0) return null;
        var s = _shipmentQueue.Dequeue();
        Console.WriteLine($"Поставка '{s}' обработана импортёром {CompanyName}.");
        return s;
    }

    public bool RegisterPortVisit(string port)
    {
        bool added = _visitedPorts.Add(port);
        if (added) Console.WriteLine($"Порт '{port}' добавлен в историю {CompanyName}.");
        return added;
    }

    public double CalculateTotalCostByCategory(double baseCost, string category)
    {
        double duty = GetCustomsDuty(category);
        return baseCost + baseCost * duty / 100;
    }

    protected override double CalculatePaymentCore(double amount) => CalculateTotalCost(amount);

    protected override double CalculateShippingCore(double weightKg) => weightKg * 25 + TransitDays * 5;
}

 
// GENERIC КЛАССЫ
 

public class SupplierRepository<T> where T : Supplier
{
    private List<T> _items = new List<T>();

    public int Count => _items.Count;

    public void Add(T supplier)
    {
        Console.WriteLine($"Поставщик {supplier.CompanyName} ({supplier.GetType().Name}) добавлен в репозиторий.");
        _items.Add(supplier);
    }

    public T GetById(int id) => _items.FirstOrDefault(s => s.SupplierId == id);
    public T GetTopRated() => _items.OrderByDescending(s => s.Rating).FirstOrDefault();
    public double GetAverageRating() => _items.Count > 0 ? _items.Average(s => s.Rating) : 0;
    public List<T> GetAll() => _items;
}

public static class SupplierHelper
{
    public static T FindBest<T>(List<T> suppliers) where T : Supplier
        => suppliers.OrderByDescending(s => s.Rating).FirstOrDefault();
}

 
// СЕРВИСЫ
 

public interface ILogger { void Log(string message); }

public class ConsoleLogger : ILogger
{
    public void Log(string message) => Console.WriteLine($"[LOG {DateTime.Now:HH:mm:ss}] {message}");
}

public interface INotificationService { void Notify(string email, string message); }

public class EmailNotificationService : INotificationService
{
    private readonly ILogger _logger;
    public EmailNotificationService(ILogger logger) { _logger = logger; }
    public void Notify(string email, string message) => _logger.Log($"Email на {email}: {message}");
}

public interface IPricingStrategy { double ApplyPricing(Supplier supplier, double baseAmount); }

public class StandardPricing : IPricingStrategy
{
    public double ApplyPricing(Supplier supplier, double baseAmount) => baseAmount;
}

public class RatingBasedPricing : IPricingStrategy
{
    public double ApplyPricing(Supplier supplier, double baseAmount)
    {
        if (supplier.Rating >= 4.5) return baseAmount * 0.95;
        if (supplier.Rating >= 4.0) return baseAmount * 0.98;
        return baseAmount;
    }
}

// Конструкторная инъекция: OrderService не создаёт зависимости сам
public class OrderService
{
    private readonly ILogger _logger;
    private readonly IPricingStrategy _pricing;
    private readonly INotificationService _notifier;

    public OrderService(ILogger logger, IPricingStrategy pricing, INotificationService notifier)
    {
        _logger = logger;
        _pricing = pricing;
        _notifier = notifier;
    }

    public bool PlaceOrder(Supplier supplier, string details, double amount, double weightKg)
    {
        _logger.Log($"Оформление заказа у {supplier.CompanyName} ({_pricing.GetType().Name})");

        if (!supplier.IsActive)
        {
            _logger.Log("Поставщик неактивен — заказ отклонён.");
            return false;
        }

        IPayable payable = supplier;
        IShippable shippable = supplier;
        double payment = payable.CalculatePayment(_pricing.ApplyPricing(supplier, amount));
        double shipping = shippable.CalculateShippingCost(weightKg);
        double total = payment + shipping;

        if (!supplier.IsCreditAvailable(total))
        {
            _logger.Log($"Сумма {total:F2} превышает кредитный лимит {supplier.CreditLimit:F0} — заказ отклонён.");
            return false;
        }

        _logger.Log($"Оплата: {payment:F2}, доставка: {shipping:F2}, итого: {total:F2}");
        Console.WriteLine(supplier.SubmitOrder(details));
        _notifier.Notify(supplier.ContactEmail, $"Ваш заказ '{details}' принят.");
        return true;
    }
}

 
// DI-КОНТЕЙНЕР
 

public class ServiceContainer
{
    private class Registration
    {
        public Type Implementation;
        public bool IsSingleton;
        public object Instance;
    }

    private readonly Dictionary<Type, Registration> _registrations = new Dictionary<Type, Registration>();
    private readonly HashSet<Type> _resolving = new HashSet<Type>();

    public void RegisterSingleton<TService, TImpl>() where TImpl : TService
        => _registrations[typeof(TService)] = new Registration { Implementation = typeof(TImpl), IsSingleton = true };

    public void RegisterTransient<TService, TImpl>() where TImpl : TService
        => _registrations[typeof(TService)] = new Registration { Implementation = typeof(TImpl), IsSingleton = false };

    public T Resolve<T>() => (T)Resolve(typeof(T));

    private object Resolve(Type serviceType)
    {
        if (!_registrations.TryGetValue(serviceType, out Registration reg))
            throw new InvalidOperationException($"Сервис {serviceType.Name} не зарегистрирован!");

        if (reg.IsSingleton && reg.Instance != null)
            return reg.Instance;

        if (!_resolving.Add(serviceType))
            throw new InvalidOperationException($"Циклическая зависимость при создании {serviceType.Name}!");

        try
        {
            var ctor = reg.Implementation.GetConstructors()
                          .OrderByDescending(c => c.GetParameters().Length)
                          .First();
            object[] args = ctor.GetParameters().Select(p => Resolve(p.ParameterType)).ToArray();
            object instance = ctor.Invoke(args);

            if (reg.IsSingleton) reg.Instance = instance;
            return instance;
        }
        finally
        {
            _resolving.Remove(serviceType);
        }
    }
}

 
// MAIN (top-level statements)
 

var manufacturer = new Manufacturer(1, "ООО 'ТехноСтрой'", "Строительные материалы", "info@technostroy.ru", 2005);
manufacturer.ProductionCapacity = 10000;
manufacturer.QualityCertificate = "ISO 9001";
manufacturer.DefectRate = 1.5;
manufacturer.LeadTimeDays = 14;
manufacturer.MinOrderQuantity = 100;
manufacturer.WarrantyMonths = 24;
manufacturer.CertificationBody = "TÜV Rheinland";
manufacturer.PaymentTermsDays = 14;
manufacturer.CreditLimit = 500000;
manufacturer.BankAccount = "40817810099910004312";
manufacturer.UpdateRating(4.7);

var manufacturer2 = new Manufacturer(4, "ООО 'МеталлПром'", "Металлопрокат", "contact@metalprom.ru", 1998);
manufacturer2.CreditLimit = 100000;
manufacturer2.UpdateRating(4.1);

var retailer = new Retailer(2, "ООО 'Мир Электроники'", "Бытовая техника", "shop@mirtech.ru", "г. Москва, ул. Тверская, 15");
retailer.EmployeeCount = 25;
retailer.OpeningHours = "09:00-21:00";
retailer.DiscountPercentage = 10;
retailer.ReturnPolicyDays = 14;
retailer.HasOnlineStore = true;
retailer.AverageCheck = 3500;
retailer.LoyaltyProgram = true;
retailer.CreditLimit = 200000;
retailer.BankAccount = "40817810099910005555";
retailer.UpdateRating(4.2);

var importer = new Importer(3, "ООО 'Глобал Трейд'", "Электронные компоненты", "sales@globaltrade.ru", "Китай");
importer.CustomsCode = "ТН ВЭД 8471";
importer.ShippingMethod = "Морской контейнер";
importer.ImportTax = 15;
importer.ImportLicenseNumber = "LIC-2024-0087";
importer.TransitDays = 35;
importer.InsuranceRate = 1.2;
importer.PortOfEntry = "Владивосток";
importer.CreditLimit = 300000;
importer.BankAccount = "40817810099910007777";
importer.UpdateRating(3.9);

Console.WriteLine("\n=== Перегрузка методов (overloading) ===\n");

Console.WriteLine(manufacturer.ProvideQuote());
Console.WriteLine(manufacturer.ProvideQuote(500));
Console.WriteLine(retailer.SubmitOrder("Заказ №1"));
Console.WriteLine(retailer.SubmitOrder("Заказ №1", DateTime.Now.AddDays(7)));

Console.WriteLine("\n=== Демонстрация полиморфизма (overriding) ===\n");

Supplier[] suppliers = { manufacturer, retailer, importer };

foreach (var s in suppliers)
{
    Console.WriteLine(s.GetCompanyInfo());
    Console.WriteLine(s.ProvideQuote());
    Console.WriteLine("---");
}

Console.WriteLine("\n=== Новые методы ===\n");

Console.WriteLine($"Мощности хватает на 8000 ед.: {manufacturer.CheckCapacity(8000)}");
Console.WriteLine($"Срок производства 2500 ед.: {manufacturer.EstimateProductionTime(2500)} дн.");
Console.WriteLine($"Заказ 50 ед. допустим: {manufacturer.IsOrderValid(50)}");
Console.WriteLine($"Гарантийные затраты на 10000: {manufacturer.CalculateWarrantyCost(10000):F2}");
Console.WriteLine($"Цена после скидки ритейлера: {retailer.ApplyDiscount(1000):F2}");
Console.WriteLine($"Возврат через 10 дней возможен: {retailer.CanReturn(10)}");
Console.WriteLine($"Итог чека на 3 позиции: {retailer.EstimateCheckTotal(3):F2}");
Console.WriteLine($"Бонус лояльности за 5000: {retailer.CalculateLoyaltyBonus(5000):F2}");
Console.WriteLine($"Стоимость с пошлиной импортёра: {importer.CalculateTotalCost(1000):F2}");
Console.WriteLine($"Страховка груза на 1000: {importer.CalculateInsurance(1000):F2}");
Console.WriteLine($"Ожидаемая доставка: {importer.EstimateDelivery(DateTime.Now):dd.MM.yyyy}");
Console.WriteLine($"Лицензия импортёра действительна: {importer.IsLicenseValid()}");
Console.WriteLine($"Лет на рынке {manufacturer.CompanyName}: {manufacturer.GetYearsOnMarket()}");

manufacturer.PassCertification("ISO 14001", "TÜV Rheinland");
manufacturer.ExtendPaymentTerms(10);
importer.UpdatePortOfEntry("Санкт-Петербург");
manufacturer.Audit("Иванов И.И.");

Console.WriteLine("\n=== Взаимодействие объектов ===\n");

importer.SupplyToManufacturer(manufacturer, "микросхемы, 500 шт.");
manufacturer.SupplyToRetailer(retailer, "партия строительных блоков");
retailer.OrderFromManufacturer(manufacturer2, "500 ед. цемента");
retailer.OrderFromImporter(importer, "партия смартфонов");

Console.WriteLine("\n=== Явная реализация интерфейсов ===\n");

foreach (var s in suppliers)
{
    IPayable payable = s;
    IShippable shippable = s;
    Console.WriteLine(payable.GetSummary());
    Console.WriteLine(shippable.GetSummary());
    Console.WriteLine($"  К оплате за 10000: {payable.CalculatePayment(10000):F2}");
    Console.WriteLine($"  Доставка 20 кг: {shippable.CalculateShippingCost(20):F2}");
    Console.WriteLine("---");
}

Console.WriteLine("\n=== Демонстрация Generic классов ===\n");

var repository = new SupplierRepository<Manufacturer>();
repository.Add(manufacturer);
repository.Add(manufacturer2);

Console.WriteLine($"Производителей в репозитории: {repository.Count}");
Console.WriteLine($"Средний рейтинг: {repository.GetAverageRating():F2}");
Console.WriteLine($"Лучший производитель: {repository.GetTopRated()?.CompanyName}");
Console.WriteLine($"Найден по ID=4: {repository.GetById(4)?.CompanyName}");
Console.WriteLine($"Лучший среди всех (generic-метод): {SupplierHelper.FindBest(suppliers.ToList()).CompanyName}");

Console.WriteLine("\n=== Управление зависимостями (DI-контейнер) ===\n");

var container = new ServiceContainer();
container.RegisterSingleton<ILogger, ConsoleLogger>();
container.RegisterSingleton<INotificationService, EmailNotificationService>();
container.RegisterTransient<IPricingStrategy, RatingBasedPricing>();
container.RegisterTransient<OrderService, OrderService>();

var orderService = container.Resolve<OrderService>();
Console.WriteLine($"Логгер — единственный экземпляр: {object.ReferenceEquals(container.Resolve<ILogger>(), container.Resolve<ILogger>())}\n");

orderService.PlaceOrder(manufacturer, "Кирпич, 2000 ед.", 120000, 4000);
orderService.PlaceOrder(retailer, "Телевизоры, 10 шт.", 300000, 150);   // превысит кредитный лимит

// Подмена зависимости без изменения кода OrderService
container.RegisterTransient<IPricingStrategy, StandardPricing>();
container.Resolve<OrderService>().PlaceOrder(importer, "Платы, 5000 шт.", 80000, 300);

 
// ДЕМОНСТРАЦИЯ: КОЛЛЕКЦИИ, ДЕЛЕГАТЫ, СОБЫТИЯ
 

Console.WriteLine("\n=== Коллекции, делегаты, события ===\n");

// --- Коллекции в базовом классе ---
manufacturer.AddTag("стройматериалы");
manufacturer.AddTag("опт");
manufacturer.AddTag("стройматериалы"); // дубликат не добавится
manufacturer.AddContact("менеджер", "petrov@technostroy.ru");
manufacturer.AddContact("бухгалтер", "buh@technostroy.ru");

manufacturer.AddCertification("ISO 9001");
manufacturer.AddCertification("ISO 9001"); // дубликат
manufacturer.AddCertification("ISO 14001");

Console.WriteLine($"Теги: {string.Join(", ", manufacturer.Tags)}");
Console.WriteLine($"Контакт 'менеджер': {manufacturer.GetContact("менеджер")}");
Console.WriteLine($"Сертификатов: {manufacturer.Certifications.Count}");

// --- Коллекции в производных ---
manufacturer.AddProductionLine("Линия A");
manufacturer.AddProductionLine("Линия B");
manufacturer.SetComponentCost("цемент", 1200);
manufacturer.SetComponentCost("песок", 300);
Console.WriteLine($"Суммарная стоимость комплектующих: {manufacturer.GetTotalComponentCost():F2}");
Console.WriteLine($"Загрузка линии при 2500 ед.: {manufacturer.EstimateLineLoad(2500):F1}%");
manufacturer.PassQualityCheck("Прочность");
manufacturer.PassQualityCheck("Прочность");

retailer.AddStock("Телевизор", 20);
retailer.SellProduct("Телевизор", 5);
retailer.AddPromotion("Чёрная пятница");
retailer.RegisterLoyaltyMember("client@mail.ru");
retailer.RegisterLoyaltyMember("client@mail.ru");
Console.WriteLine($"Кэшбэк с 5000: {retailer.CalculateCashback(5000):F2}");

importer.SetCustomsDuty("Электроника", 20);
importer.EnqueueShipment("Партия №1");
importer.EnqueueShipment("Партия №2");
importer.RegisterPortVisit("Владивосток");
importer.RegisterPortVisit("Санкт-Петербург");
importer.ProcessNextShipment();
Console.WriteLine($"Стоимость с пошлиной (Электроника): {importer.CalculateTotalCostByCategory(1000, "Электроника"):F2}");

// --- События ---
manufacturer.RatingChanged += (s, oldR, newR) =>
    Console.WriteLine($"[EVENT] Рейтинг {s.CompanyName}: {oldR} -> {newR}");

retailer.OrderPlaced += (s, details, total) =>
    Console.WriteLine($"[EVENT] Заказ у {s.CompanyName}: '{details}'");

importer.SupplierDeactivated += (s, reason) =>
    Console.WriteLine($"[EVENT] {s.CompanyName} деактивирован: {reason}");

manufacturer.UpdateRating(4.9);      // вызовет RatingChanged
retailer.SubmitOrder("Заказ №2");     // вызовет OrderPlaced
importer.Deactivate();                // вызовет SupplierDeactivated

// --- Делегаты ---
var activeAndHighRated = Supplier.FilterSuppliers(suppliers,
    s => s.IsActive && s.Rating >= 4.0);
Console.WriteLine($"Активных с рейтингом ≥ 4.0: {activeAndHighRated.Count}");

var names = Supplier.ProjectSuppliers(suppliers, s => $"{s.CompanyName} [{s.Rating}]");
Console.WriteLine($"Проекция: {string.Join(" | ", names)}");

DiscountCalculator bulkDiscount = (price, s) => price > 10000 ? price * 0.9 : price;
Console.WriteLine($"Цена 15000 со скидкой: {manufacturer.CalculateDiscountedPrice(15000, bulkDiscount):F2}");
Console.WriteLine($"Цена 5000 со скидкой: {manufacturer.CalculateDiscountedPrice(5000, bulkDiscount):F2}");
```
