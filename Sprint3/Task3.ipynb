```
using System;
using System.Collections.Generic;
using System.Linq;

public class Supplier
{
    // Приватные поля
    private int _supplierId;
    private string _companyName;
    private string _productType;
    private double _rating;
    private bool _isActive;
    private string _contactEmail;
    private int _totalOrders;

    // Публичные свойства с геттерами и сеттерами
    public int SupplierId
    {
        get { return _supplierId; }
        set { _supplierId = value; }
    }

    public string CompanyName
    {
        get { return _companyName; }
        set
        {
            if (!string.IsNullOrEmpty(value))
                _companyName = value;
            else
                throw new ArgumentException("Название компании не может быть пустым!");
        }
    }

    public string ProductType
    {
        get { return _productType; }
        protected set { _productType = value; } // Защищённый сеттер для наследников
    }

    public double Rating
    {
        get { return _rating; }
        protected set
        {
            if (value >= 0 && value <= 5)
                _rating = value;
            else
                throw new ArgumentException("Рейтинг должен быть в диапазоне от 0 до 5!");
        }
    }

    public bool IsActive
    {
        get { return _isActive; }
        set { _isActive = value; }
    }

    public string ContactEmail
    {
        get { return _contactEmail; }
        set
        {
            if (value != null && value.Contains("@"))
                _contactEmail = value;
            else
                throw new ArgumentException("Некорректный email адрес!");
        }
    }

    public int TotalOrders
    {
        get { return _totalOrders; }
        private set { _totalOrders = value; }
    }

    // Конструктор
    public Supplier(int supplierId, string companyName, string productType, string contactEmail)
    {
        SupplierId = supplierId;
        CompanyName = companyName;
        ProductType = productType;
        ContactEmail = contactEmail;
        Rating = 0;
        IsActive = true;
        TotalOrders = 0;
    }

    // Virtual методы для переопределения (overriding) 
    public virtual string GetCompanyInfo()
    {
        return $"ID: {SupplierId}, Компания: {CompanyName}, Тип продукции: {ProductType}, Рейтинг: {Rating}";
    }

    // Перегрузка (overloading): два метода ProvideQuote с разной сигнатурой 
    public virtual string ProvideQuote()
    {
        return $"Котировка от {CompanyName}: стандартные условия поставки {ProductType}.";
    }

    public virtual string ProvideQuote(int quantity)
    {
        return $"Котировка от {CompanyName} на партию {quantity} ед. {ProductType}.";
    }

    public virtual void SendQuote()
    {
        Console.WriteLine(ProvideQuote());
    }

    // ---- Перегрузка (overloading): два метода SubmitOrder с разной сигнатурой ----
    public virtual string SubmitOrder(string orderDetails)
    {
        RegisterOrder();
        return $"Заказ '{orderDetails}' отправлен поставщику {CompanyName} (ID: {SupplierId}).";
    }

    public virtual string SubmitOrder(string orderDetails, DateTime deliveryDate)
    {
        RegisterOrder();
        return $"Заказ '{orderDetails}' отправлен поставщику {CompanyName}, ожидаемая дата доставки: {deliveryDate:dd.MM.yyyy}.";
    }

    // Метод для взаимодействия с другими поставщиками
    public void LinkWithSupplier(Supplier other)
    {
        Console.WriteLine($"Поставщик {CompanyName} (ID: {SupplierId}) связан с поставщиком {other.CompanyName} (ID: {other.SupplierId})");
    }

    // Новые методы
    public void UpdateRating(double newRating)
    {
        Rating = newRating;
        Console.WriteLine($"Рейтинг {CompanyName} обновлён: {Rating}");
    }

    public void Deactivate()
    {
        IsActive = false;
        Console.WriteLine($"{CompanyName} деактивирован.");
    }

    protected void RegisterOrder()
    {
        TotalOrders++;
    }
}


public class Manufacturer : Supplier
{
    private int _foundedYear;
    private int _productionCapacity;
    private string _qualityCertificate;
    private double _defectRate;

    public int FoundedYear
    {
        get { return _foundedYear; }
        set
        {
            if (value > 1800 && value <= DateTime.Now.Year)
                _foundedYear = value;
            else
                throw new ArgumentException("Некорректный год основания!");
        }
    }

    public int ProductionCapacity
    {
        get { return _productionCapacity; }
        set { _productionCapacity = value; }
    }

    public string QualityCertificate
    {
        get { return _qualityCertificate; }
        set { _qualityCertificate = value; }
    }

    public double DefectRate
    {
        get { return _defectRate; }
        set
        {
            if (value >= 0 && value <= 100)
                _defectRate = value;
            else
                throw new ArgumentException("Процент брака должен быть от 0 до 100!");
        }
    }

    public Manufacturer(int supplierId, string companyName, string productType, string contactEmail, int foundedYear)
        : base(supplierId, companyName, productType, contactEmail)
    {
        FoundedYear = foundedYear;
    }

    // Переопределение (overriding) обоих перегруженных вариантов ProvideQuote
    public override string ProvideQuote()
    {
        return $"Котировка от производителя {CompanyName} (основан в {FoundedYear} году): " +
               $"специальные условия для {ProductType}.";
    }

    public override string ProvideQuote(int quantity)
    {
        string note = quantity > ProductionCapacity ? " (требуется доп. время на изготовление)" : "";
        return $"Котировка от производителя {CompanyName} на {quantity} ед. {ProductType}{note}.";
    }

    public string GetManufacturerDetails()
    {
        return $"Производитель {CompanyName}, основан в {FoundedYear} году, " +
               $"мощность: {ProductionCapacity} ед./мес, сертификат: {QualityCertificate}, брак: {DefectRate}%.";
    }

    public void SupplyToRetailer(Retailer other, string productDetails)
    {
        Console.WriteLine($"Производитель {CompanyName} поставляет товар '{productDetails}' ритейлеру {other.CompanyName}");
        Console.WriteLine($"Магазин получателя: {other.StoreLocation}");
    }

    public bool CheckCapacity(int requestedQuantity)
    {
        return requestedQuantity <= ProductionCapacity;
    }
}

public class Retailer : Supplier
{
    private string _storeLocation;
    private int _employeeCount;
    private string _openingHours;
    private double _discountPercentage;

    public string StoreLocation
    {
        get { return _storeLocation; }
        set
        {
            if (!string.IsNullOrEmpty(value))
                _storeLocation = value;
            else
                throw new ArgumentException("Местоположение магазина не может быть пустым!");
        }
    }

    public int EmployeeCount
    {
        get { return _employeeCount; }
        set { _employeeCount = value; }
    }

    public string OpeningHours
    {
        get { return _openingHours; }
        set { _openingHours = value; }
    }

    public double DiscountPercentage
    {
        get { return _discountPercentage; }
        set
        {
            if (value >= 0 && value <= 100)
                _discountPercentage = value;
            else
                throw new ArgumentException("Скидка должна быть от 0 до 100%!");
        }
    }

    public Retailer(int supplierId, string companyName, string productType, string contactEmail, string storeLocation)
        : base(supplierId, companyName, productType, contactEmail)
    {
        StoreLocation = storeLocation;
    }

    // Переопределение (overriding) обоих перегруженных вариантов SubmitOrder
    public override string SubmitOrder(string orderDetails)
    {
        RegisterOrderInternal();
        return $"Заказ '{orderDetails}' отправлен ритейлеру {CompanyName} " +
               $"(магазин: {StoreLocation}, ID: {SupplierId}).";
    }

    public override string SubmitOrder(string orderDetails, DateTime deliveryDate)
    {
        RegisterOrderInternal();
        return $"Заказ '{orderDetails}' отправлен ритейлеру {CompanyName} (магазин: {StoreLocation}), " +
               $"ожидается к {deliveryDate:dd.MM.yyyy}.";
    }

    private void RegisterOrderInternal()
    {
        // пример переиспользования защищённого метода базового класса
    }

    public string GetRetailerInfo()
    {
        return $"Ритейлер {CompanyName}, расположение: {StoreLocation}, сотрудников: {EmployeeCount}, " +
               $"часы работы: {OpeningHours}, скидка: {DiscountPercentage}%.";
    }

    public void OrderFromManufacturer(Manufacturer other, string orderDetails)
    {
        Console.WriteLine($"Ритейлер {CompanyName} оформляет заказ '{orderDetails}' у производителя {other.CompanyName}");
        Console.WriteLine(other.SubmitOrder(orderDetails));
    }

    public void OrderFromImporter(Importer other, string orderDetails)
    {
        Console.WriteLine($"Ритейлер {CompanyName} оформляет заказ '{orderDetails}' у импортёра {other.CompanyName}");
        Console.WriteLine(other.SubmitOrder(orderDetails));
    }

    public double ApplyDiscount(double price)
    {
        return price - price * DiscountPercentage / 100;
    }
}


public class Importer : Supplier
{
    private string _originCountry;
    private string _customsCode;
    private string _shippingMethod;
    private double _importTax;

    public string OriginCountry
    {
        get { return _originCountry; }
        set { _originCountry = value; }
    }

    public string CustomsCode
    {
        get { return _customsCode; }
        set
        {
            if (!string.IsNullOrEmpty(value))
                _customsCode = value;
            else
                throw new ArgumentException("Таможенный код не может быть пустым!");
        }
    }

    public string ShippingMethod
    {
        get { return _shippingMethod; }
        set { _shippingMethod = value; }
    }

    public double ImportTax
    {
        get { return _importTax; }
        set
        {
            if (value >= 0 && value <= 100)
                _importTax = value;
            else
                throw new ArgumentException("Пошлина должна быть от 0 до 100%!");
        }
    }

    public Importer(int supplierId, string companyName, string productType, string contactEmail, string originCountry)
        : base(supplierId, companyName, productType, contactEmail)
    {
        OriginCountry = originCountry;
    }

    public override string GetCompanyInfo()
    {
        return base.GetCompanyInfo() + $", Страна происхождения: {OriginCountry}, пошлина: {ImportTax}%";
    }

    public string GetImportDetails()
    {
        return $"Импортер {CompanyName} поставляет {ProductType} из {OriginCountry}, " +
               $"способ доставки: {ShippingMethod}, таможенный код: {CustomsCode}.";
    }

    public void SupplyToManufacturer(Manufacturer other, string componentDetails)
    {
        Console.WriteLine($"Импортёр {CompanyName} поставляет компоненты '{componentDetails}' производителю {other.CompanyName}");
        Console.WriteLine($"Страна происхождения: {OriginCountry}");
    }

    public double CalculateTotalCost(double baseCost)
    {
        return baseCost + baseCost * ImportTax / 100;
    }
}

// любой конкретный тип поставщика (Manufacturer, Retailer, Importer)
public class SupplierRepository<T> where T : Supplier
{
    private List<T> _items = new List<T>();

    public void Add(T supplier)
    {
        _items.Add(supplier);
    }

    public T GetById(int id)
    {
        return _items.FirstOrDefault(s => s.SupplierId == id);
    }

    public List<T> GetAll()
    {
        return _items;
    }

    public int Count => _items.Count;

    public T GetTopRated()
    {
        return _items.OrderByDescending(s => s.Rating).FirstOrDefault();
    }

    public double GetAverageRating()
    {
        return _items.Count > 0 ? _items.Average(s => s.Rating) : 0;
    }
}

// Статический класс с generic-методом общего назначения
public static class SupplierHelper
{
    public static T FindBest<T>(List<T> suppliers) where T : Supplier
    {
        return suppliers.OrderByDescending(s => s.Rating).FirstOrDefault();
    }
}
Manufacturer myManufacturer = new Manufacturer(1, "ООО 'ТехноСтрой'", "Строительные материалы", "info@technostroy.ru", 2005);
Manufacturer myManufacturer2 = new Manufacturer(4, "ООО 'МеталлПром'", "Металлопрокат", "contact@metalprom.ru", 1998);

Retailer myRetailer = new Retailer(2, "ООО 'Мир Электроники'", "Бытовая техника", "shop@mirtech.ru", "г. Москва, ул. Тверская, 15");
Importer myImporter = new Importer(3, "ООО 'Глобал Трейд'", "Электронные компоненты", "sales@globaltrade.ru", "Китай");

// Настройка дополнительных атрибутов
myManufacturer.ProductionCapacity = 10000;
myManufacturer.QualityCertificate = "ISO 9001";
myManufacturer.DefectRate = 1.5;
myManufacturer.UpdateRating(4.7);

myRetailer.EmployeeCount = 25;
myRetailer.OpeningHours = "09:00-21:00";
myRetailer.DiscountPercentage = 10;
myRetailer.UpdateRating(4.2);

myImporter.CustomsCode = "ТН ВЭД 8471";
myImporter.ShippingMethod = "Морской контейнер";
myImporter.ImportTax = 15;
myImporter.UpdateRating(3.9);

Console.WriteLine("=== Перегрузка методов (overloading) ===\n");
// Два варианта ProvideQuote: без параметров и с указанием количества
Console.WriteLine(myManufacturer.ProvideQuote());
Console.WriteLine(myManufacturer.ProvideQuote(500));
Console.WriteLine();

// Два варианта SubmitOrder: без даты и с датой доставки
Console.WriteLine(myRetailer.SubmitOrder("Заказ №1"));
Console.WriteLine(myRetailer.SubmitOrder("Заказ №1", DateTime.Now.AddDays(7)));
Console.WriteLine();

Console.WriteLine("=== Переопределение методов (overriding) — полиморфизм ===\n");
// Один и тот же вызов через базовый тип даёт разное поведение
List<Supplier> allSuppliers = new List<Supplier> { myManufacturer, myRetailer, myImporter };
foreach (Supplier s in allSuppliers)
{
    Console.WriteLine(s.ProvideQuote());       // вызовется переопределённая версия
    Console.WriteLine(s.GetCompanyInfo());
    Console.WriteLine();
}

Console.WriteLine("=== Новые методы ===\n");
Console.WriteLine(myManufacturer.CheckCapacity(8000) ? "Производитель справится с объёмом 8000 ед." : "Недостаточно мощностей.");
Console.WriteLine($"Цена после скидки ритейлера: {myRetailer.ApplyDiscount(1000):F2}");
Console.WriteLine($"Итоговая стоимость с учётом пошлины импортёра: {myImporter.CalculateTotalCost(1000):F2}");
Console.WriteLine();

Console.WriteLine("=== Взаимодействие объектов ===\n");
myImporter.SupplyToManufacturer(myManufacturer, "микросхемы, 500 шт.");
myManufacturer.SupplyToRetailer(myRetailer, "партия строительных блоков");
myRetailer.OrderFromManufacturer(myManufacturer2, "500 ед. цемента");
myRetailer.OrderFromImporter(myImporter, "партия смартфонов");
Console.WriteLine();

Console.WriteLine("=== Generic-класс SupplierRepository<T> ===\n");
SupplierRepository<Manufacturer> manufacturerRepo = new SupplierRepository<Manufacturer>();
manufacturerRepo.Add(myManufacturer);
manufacturerRepo.Add(myManufacturer2);

Console.WriteLine($"Производителей в репозитории: {manufacturerRepo.Count}");
Console.WriteLine($"Средний рейтинг производителей: {manufacturerRepo.GetAverageRating():F2}");
Manufacturer top = manufacturerRepo.GetTopRated();
Console.WriteLine($"Лучший производитель: {top?.CompanyName} (рейтинг {top?.Rating})");
Manufacturer found = manufacturerRepo.GetById(4);
Console.WriteLine($"Найден по ID=4: {found?.CompanyName}");
Console.WriteLine();

// Generic-метод, работающий с любым списком поставщиков
Supplier bestOverall = SupplierHelper.FindBest(allSuppliers);
Console.WriteLine($"Лучший поставщик среди всех (generic-метод): {bestOverall.CompanyName}, рейтинг {bestOverall.Rating}");
```
