<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Supplier в C#, который будет представлять информацию о
поставщиках товаров или услуг. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;


    // Базовый класс Supplier
    public class Supplier
    {
        // Атрибуты
        public int SupplierId { get; set; }
        public string CompanyName { get; set; }
        public string ProductType { get; set; }

        // Конструкторы
        public Supplier() { }

        public Supplier(int supplierId, string companyName, string productType)
        {
            SupplierId = supplierId;
            CompanyName = companyName;
            ProductType = productType;
        }

        // Виртуальный метод получения информации о компании
        public virtual string GetCompanyInfo()
        {
            return $"ID: {SupplierId}, Компания: {CompanyName}, Тип продукции: {ProductType}";
        }

        // Виртуальный метод предоставления котировки
        public virtual string ProvideQuote()
        {
            return $"Котировка от {CompanyName}: стандартные условия поставки {ProductType}.";
        }

        // Виртуальный метод отправки заказа
        public virtual string SubmitOrder(string orderDetails)
        {
            return $"Заказ '{orderDetails}' отправлен поставщику {CompanyName} (ID: {SupplierId}).";
        }
    }

    // Производный класс - Производитель (Manufacturer)
    public class Manufacturer : Supplier
    {
        // Дополнительный атрибут
        public int FoundedYear { get; set; }

        // Конструкторы
        public Manufacturer() { }

        public Manufacturer(int supplierId, string companyName, string productType, int foundedYear)
            : base(supplierId, companyName, productType)
        {
            FoundedYear = foundedYear;
        }

        // Переопределение метода ProvideQuote() с добавлением года основания
        public override string ProvideQuote()
        {
            return $"Котировка от производителя {CompanyName} (основан в {FoundedYear} году): " +
                   $"специальные условия для {ProductType} с учетом производственных мощностей.";
        }

        // Дополнительный метод (демонстрация расширения функциональности)
        public string GetManufacturerDetails()
        {
            return $"Производитель {CompanyName}, основан в {FoundedYear} году, производит {ProductType}.";
        }
    }

    // Производный класс - Ритейлер (Retailer)
    public class Retailer : Supplier
    {
        // Дополнительный атрибут
        public string StoreLocation { get; set; }

        // Конструкторы
        public Retailer() { }

        public Retailer(int supplierId, string companyName, string productType, string storeLocation)
            : base(supplierId, companyName, productType)
        {
            StoreLocation = storeLocation;
        }

        // Переопределение метода SubmitOrder() с добавлением расположения магазина
        public override string SubmitOrder(string orderDetails)
        {
            return $"Заказ '{orderDetails}' отправлен ритейлеру {CompanyName} " +
                   $"(магазин находится по адресу: {StoreLocation}, ID: {SupplierId}).";
        }

        // Дополнительный метод (демонстрация расширения функциональности)
        public string GetRetailerInfo()
        {
            return $"Ритейлер {CompanyName}, расположение: {StoreLocation}, ассортимент: {ProductType}.";
        }
    }

    // Производный класс - Импортер (Importer)
    public class Importer : Supplier
    {
        // Дополнительный атрибут
        public string OriginCountry { get; set; }

        // Конструкторы
        public Importer() { }

        public Importer(int supplierId, string companyName, string productType, string originCountry)
            : base(supplierId, companyName, productType)
        {
            OriginCountry = originCountry;
        }

        // Переопределение метода GetCompanyInfo() с добавлением страны происхождения
        public override string GetCompanyInfo()
        {
            return $"Импортер: ID: {SupplierId}, Компания: {CompanyName}, " +
                   $"Тип продукции: {ProductType}, Страна происхождения: {OriginCountry}";
        }

        // Дополнительный метод (демонстрация расширения функциональности)
        public string GetImportDetails()
        {
            return $"Импортер {CompanyName} поставляет {ProductType} из {OriginCountry}.";
        }
    }

    // Программа для демонстрации
    class Program
    {
        static void Main(string[] args)
        {
            Console.WriteLine("=== Демонстрация наследования и полиморфизма ===\n");

            // Создание объектов различных типов
            Supplier supplier = new Supplier(1, "ООО 'Стандарт'", "Офисная мебель");
            Manufacturer manufacturer = new Manufacturer(2, "ООО 'ТехноСтрой'", "Строительные материалы", 2005);
            Retailer retailer = new Retailer(3, "ООО 'Мир Электроники'", "Бытовая техника", "г. Москва, ул. Тверская, 15");
            Importer importer = new Importer(4, "ООО 'Глобал Трейд'", "Электронные компоненты", "Китай");

            // ========== ДЕМОНСТРАЦИЯ 1: Массив базового типа (полиморфизм) ==========
            Console.WriteLine("--- Демонстрация 1: Полиморфизм через массив Supplier[] ---");
            Supplier[] suppliers = { supplier, manufacturer, retailer, importer };

            foreach (var s in suppliers)
            {
                Console.WriteLine(s.GetCompanyInfo());
                Console.WriteLine(s.ProvideQuote());
                Console.WriteLine(s.SubmitOrder("Тестовый заказ №123"));
                Console.WriteLine(new string('-', 60));
            }

            // ========== ДЕМОНСТРАЦИЯ 2: Специализированные методы ==========
            Console.WriteLine("\n--- Демонстрация 2: Специализированные методы производных классов ---");
            Console.WriteLine(manufacturer.GetManufacturerDetails());
            Console.WriteLine(retailer.GetRetailerInfo());
            Console.WriteLine(importer.GetImportDetails());

            // ========== ДЕМОНСТРАЦИЯ 3: Сравнение поведения переопределенных методов ==========
            Console.WriteLine("\n--- Демонстрация 3: Сравнение переопределенных методов ---");
            
            Console.WriteLine("\n[GetCompanyInfo]");
            Console.WriteLine($"  Базовый:    {supplier.GetCompanyInfo()}");
            Console.WriteLine($"  Импортер:   {importer.GetCompanyInfo()}");

            Console.WriteLine("\n[ProvideQuote]");
            Console.WriteLine($"  Базовый:        {supplier.ProvideQuote()}");
            Console.WriteLine($"  Производитель:  {manufacturer.ProvideQuote()}");

            Console.WriteLine("\n[SubmitOrder]");
            Console.WriteLine($"  Базовый:    {supplier.SubmitOrder("Заказ А")}");
            Console.WriteLine($"  Ритейлер:   {retailer.SubmitOrder("Заказ Б")}");

            // ========== ДЕМОНСТРАЦИЯ 4: Приведение типов ==========
            Console.WriteLine("\n--- Демонстрация 4: Приведение типов (casting) ---");
            
            // Приведение к базовому типу (упаковка)
            Supplier supplierAsManufacturer = manufacturer;
            Console.WriteLine("Производитель как Supplier:");
            Console.WriteLine($"  {supplierAsManufacturer.GetCompanyInfo()}");
            Console.WriteLine($"  {supplierAsManufacturer.ProvideQuote()}");

            // Обратное приведение (распаковка)
            if (supplierAsManufacturer is Manufacturer castedManufacturer)
            {
                Console.WriteLine("\nОбратное приведение к Manufacturer:");
                Console.WriteLine($"  {castedManufacturer.GetManufacturerDetails()}");
            }

            Console.WriteLine("\nНажмите любую клавишу для выхода...");
            Console.ReadKey();
        }
    }
```
